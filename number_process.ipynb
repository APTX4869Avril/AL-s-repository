对特定轴进行求和

```python
x = torch.arange(24, dtype=torch.float32).reshape(2,3,4)
x_sum_axis0 = x.sum(axis=0) # 对x的第0轴求和(以第0轴的向量整体求和)
x_sum_axis1 = x.sum(axis=1) # 对x的第1轴求和
x_sum_axis01 = x.sum(axis=(0,1)) # 对第0轴和第1轴求和, 先对第0轴求和，再对第1轴求和
x, x_sum_axis0, x_sum_axis1, x_sum_axis01
```

```python
x.cumsum(axis=0) # 对x的第0轴求和(以第0轴的向量整体求和，[0，x, x]不变，[1，x, x]累加)
```

对特定轴/维度求均值

```python
x_mean = x.mean() # 对x求均值
x_mean_axis0 = x.mean(axis=0) # 对x的第0轴求均值,x.mean(axis=0)=x.sum(axis=0)/x.shape[0]
x, x_mean_axis0
```

计算总值和均值维度不变

```python
x_sum_axis1 = x.sum(axis=1, keepdim=True) # 对x的第1轴求和, keepdim=True表示保持原有的维度
x_sum_axis1.shape, x_sum_axis1 # x_sum_axis1的形状是(2,1,4), x的形状是(2,3,4), 使可以继续进行广播
```

处理张量

```python
X = torch.arange(12,dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0,3,4,7],[3,4,5,6],[1,7,3,2]])
torch.cat((X, Y), dim=0),torch.cat((X, Y), dim=1)
```

```python
X == Y # 布尔值运算
```

元素索引和矩阵特定赋值

```python
X[0:3],X[-1] # -1可以用来访问张量最后一行，0:3访问前三行
```

```python
X[2,3] = 9 # 用来对第三行，第四列的数字进行重写
X
```

```python
X[:2, :] = 12 #用来对多个元素赋值相同值
X
```

两个张量形状不同，仍然可以使用广播机制

```python
a = torch.arange(3).reshape(3,1)
b = torch.tensor([1, 2])
a, b
a + b
```

运行一些操作，可能会导致为新结果分配内存

```python
before = id(Y)
Y = X + Y
id(Y) == before
```

执行原地操作

```python
Z = torch.zeros_like(Y) # 形成 shape和数据类型与 Y一致的张量
print('id(Z)=',id(Z))
Z[:] = X + Y
print('id(Z)=',id(Z))
```

```python
before = id(Z)
Z += Y    # 使用该方法可以不改变 Z的位置，减少操作内存的开销
id(Z) == before
```

给定具有相同格式的张量，按任何元素二元计算的结果都是同样格式的张量

```python
A = torch.arange(20, dtype=torch.float32).reshape(5,4)
B = A.clone()
print(A, A+B)
print(A*B) # 逐元素相乘, 内积
A == B # 判断两个张量是否相等
```

➡️ 数据预处理

创建一个人工数据集，并储存在csv文件中

```python
import os

os.makedirs(os.path.join('..', 'data'), exist_ok=True)
data_file = (os.path.join('..', 'data', 'house_tiny.csv'))
with open (data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n') # 列名
    f.write('NA,Pave,172500\n') #每行表示一个数据样本，NA表示缺失值
    f.write('2,NA,127500\n')
    f.write('4,NA,178900\n')
    f.write('NA,NA,140000\n')
```

从创建的csv文件中读取原始数据集

```python
import pandas as pd
data = pd.read_csv(data_file) #read_csv 可以用于读取csv文件数据
print(data)
data
```

为了处理缺失数据，我们可以采用 插值和删除的方法

```python
inputs , outputs = data.iloc[:, 0:2], data.iloc[:, 2] #iloc为index location, 用于索引
inputs = inputs.fillna(inputs.mean(numeric_only=True))
print(inputs)
```

对于inputs中的类别值或离散值，我们将NAN视为一种类别值

```python
inputs = pd.get_dummies(inputs, dummy_na=True) # get_dummies 将类别特征转换为数值特征
print(inputs)
```

现在inputs和outputs都是数值，则可以转为张量

```python
# 确保 inputs 和 outputs 是数值类型
import torch
inputs = inputs.astype('float32') # astype 用于转换数据类型
# 转换为pytorch张量
x, y = torch.tensor(inputs.values), torch.tensor(outputs.values)
x, y
```
